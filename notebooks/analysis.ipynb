```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
df = pd.read_csv("../data/processed/labeled_features.csv")
df.head()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df["avg_weekly_commits"],bins=10)
plt.title("Average Weekly Commits")
plt.xlabel("Commits per week")
plt.show()
```

```python
plt.figure()
sns.histplot(df["late_night_commit_cent"], bins=10)
plt.title("Late Night Commits Percentage")
plt.xlabel("Late Night Commits (%)")
plt.show()
```

```python
plt.figure()
sns.histplot(df["commit_irregularity"], bins=10)
plt.title("Commit Irregularity (Std Dev)")
plt.xlabel("Irregularity")
plt.show()
```

```python
df["burnout_risk"].value_counts()
```

```python
plt.figure()
df["burnout_risk"].value_counts().plot(kind='bar')
plt.title("Burnout Risk Distribution")
plt.xlabel("Burnout Risk")
plt.ylabel("Number of Developers")
plt.show()
```
